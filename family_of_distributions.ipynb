
Extracting random variables from bernoulli sequences
- Uniform to Bernoulli
    - Bernoulli to Binomial
        - Binomial to Poisson
        - Binomial to Normal
        - Binomial to Exponential
        - Binomial to Geometric


```python
import random

random.seed(34) # for reproducibility

def generate_bernoulli_sequence(N, p):
    arr = []
    for _ in range(N):
        if random.random() < p:
            arr.append(1)
        else:
            arr.append(0)
    return arr

# Example usage:
N = 20
p = 0.2
sequence = generate_bernoulli_sequence(N, p)
print(sequence)
```
